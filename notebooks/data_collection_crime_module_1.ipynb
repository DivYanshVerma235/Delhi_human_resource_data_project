```python
import pandas as pd
import numpy as np
import sqlite3 as sql
data=pd.read_csv('raw data/IPC district wise crime rate.csv',index_col=0)
data.head(10)


```

```python
data_delhi=data.loc[data.state_name=='Delhi']
data_delhi=data_delhi.drop(columns=['state_code','district_code', 'registration_circles','state_name','year',])
data_delhi



```

```python
data_delhi['district_name'].value_counts()
```

```python
district_delhi=data_delhi.groupby('district_name').sum()
district_delhi['total_crimes']=district_delhi.sum(axis=1)
print(district_delhi.columns)
district_delhi
property_crime = ["robbery","dacoity","mkng_preparation_and_assembly_for_comm_dacoity","extortion",'criminal_trespass_or_burglary',"theft"]
violent = ["murder","attempt_to_commit_murder","culpable_homicide_not_amounting_to_murder","attempt_to_commit_culpable_homicide","rape","attempt_to_commit_rape","kidnapping_and_abduction","human_trafficking"]
women_crime = ["rape","attempt_to_commit_rape","assault_on_women","cruelty_by_husband_or_his_relatives"]
social = ["offences_promoting_enmity_between_diff_grp","unnatural_offence"]
public_safety = ["causing_death_by_negligence","incidence_of_rash_driving"]
district_delhi['property_crime']=district_delhi[property_crime].sum(axis=1)
district_delhi['violent_crime']=district_delhi[violent].sum(axis=1)
district_delhi['women_crime']=district_delhi[women_crime].sum(axis=1) 
district_delhi['social_crime']=district_delhi[social].sum(axis=1)
district_delhi['public_safety']=district_delhi[public_safety].sum(axis=1)
```

```python
df = district_delhi[
    [
        'property_crime',
        'violent_crime',
        'women_crime',
        'social_crime',
        'public_safety',
        'total_crimes'
    ]
].reset_index()
numeric_cols=df.columns.drop("district_name")
df.loc[6, numeric_cols]+=df.loc[7 , numeric_cols]
df.loc[6,"district_name"]= "South"
df=df.drop(index=7).reset_index(drop=True)
df
```

```python

df.to_csv("crime_module_1.csv",index=False)
```
